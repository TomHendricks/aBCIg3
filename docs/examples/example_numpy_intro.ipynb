## Overview

NumPy is the fundamental package for scientific computing in Python. In this example, we'll cover:

- Creating arrays
- Basic operations
- Array indexing
- Simple statistics

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt

print(f"NumPy version: {np.__version__}")
```

## Creating Arrays

NumPy arrays can be created from Python lists or using built-in functions.

```python
# Create an array from a list
arr = np.array([1, 2, 3, 4, 5])
print("Array:", arr)
print("Shape:", arr.shape)
print("Data type:", arr.dtype)
```

```python
# Create arrays with built-in functions
zeros = np.zeros(5)
ones = np.ones(5)
range_arr = np.arange(0, 10, 2)

print("Zeros:", zeros)
print("Ones:", ones)
print("Range:", range_arr)
```

## Basic Operations

NumPy supports element-wise operations on arrays.

```python
# Element-wise operations
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Addition:", a + b)
print("Multiplication:", a * b)
print("Squared:", a ** 2)
```

## Generating Random Data

Let's create some random data and visualize it.

```python
# Generate random data
np.random.seed(42)  # For reproducibility
data = np.random.randn(100)

# Plot the data
plt.figure(figsize=(10, 4))
plt.plot(data)
plt.title("Random Data")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)
plt.show()
```

## Basic Statistics

NumPy provides functions for statistical analysis.

```python
# Calculate statistics
print(f"Mean: {data.mean():.2f}")
print(f"Standard deviation: {data.std():.2f}")
print(f"Minimum: {data.min():.2f}")
print(f"Maximum: {data.max():.2f}")
```
